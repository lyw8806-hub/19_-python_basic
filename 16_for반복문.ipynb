```python
#for와 range 사용하기
#for 변수 in range():
    #반복할코드

for i in range(100):
    print("hello, world")
```

```python
# 반복문에서 변수의 변화 알아보기

for i in range(10):
    print("hello, world,1")
```

```python
#시작하는 숫자와 끝나는 숫자 지정하기

for a in range(5,12):
    print("hello",a)
```

```python
for b in range(0,10,2):
    print("hello",b)
```

```python
# 숫자를 감소 시키기

for i in range(10,0):
    print(i)
```

```python

for i in range(10,0,-1):
    print(i)
```

```python
a = [1,3,5,7,9]
list(reversed(a))
```

```python
for i in reversed(range(10)):
    print(i)
```

```python
#시퀀스 객체로 반복하기
a = [10,20,30,40,50]

for num in a:
    print(num)
```

```python
for letter in "python":
    print(letter)
```
