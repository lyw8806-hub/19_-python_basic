```python
#중첩루프 사용하기
for i in range(5):
    for j  range(5):
        print("j:",j, sep="", end=" ")
    print("i:",i, "\\n", sep="")
```

```python
#사각형으로 별 출력하기
for i in range(5):
    for j in range(5):
        print("*", end="")
    print()
```

```python
#계단식으로 별 출력하기
for i in range(5):
    for j in range(5):
        if j <=i:
            print('*',end="")
        print()
```
